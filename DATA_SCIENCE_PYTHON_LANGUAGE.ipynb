HW  
Find another list manipulation function(5)   




```
numbers = [10, 20, 30, 40]

print("Original List:", numbers)

#append()
numbers.append(50)
print("After append(50):", numbers)

#extend()
numbers.extend([60, 70])
print("After extend([60, 70]):", numbers)

#insert()
numbers.insert(2, 25)
print("After insert(2, 25):", numbers)

#remove()
numbers.remove(30)
print("After remove(30):", numbers)

#pop()
removed = numbers.pop()
print("After pop():", numbers)
print("Removed element:", removed)

#index()
print("Index of 40:", numbers.index(40))

#count()
numbers.append(20)
print("Count of 20:", numbers.count(20))

#sort()
numbers.sort()
print("After sort():", numbers)

#reverse()
numbers.reverse()
print("After reverse():", numbers)

#copy()
new_list = numbers.copy()
print("Copied List:", new_list)

#Built-in functions
print("Length:", len(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))

#clear()
numbers.clear()
print("After clear():", numbers)
```

```
student = {
    "name": "Alice",
    "age": 20,
    "course": "Python"
}

print("Original Dictionary:", student)

# get()
print("Name:", student.get("name"))

# keys()
print("Keys:", student.keys())

# values()
print("Values:", student.values())

# items()
print("Items:", student.items())

# update()
student.update({"age": 21, "city": "Delhi"})
print("After update:", student)

# setdefault()
student.setdefault("grade", "A")
print("After setdefault:", student)

# pop()
student.pop("course")
print("After pop:", student)

# popitem()
student.popitem()
print("After popitem:", student)

# copy()
new_student = student.copy()
print("Copied Dictionary:", new_student)

# clear()
student.clear()
print("After clear:", student)
```

```
n = int(input("Enter the number: "))
t = n                   # Store the original number
p = len(str(n))         # Count the number of digits
s = 0

while n>0:
  d=n%10
  s+=d**p
  n=n//10

if t==s:
  print("Armstrong")
else:
  print("Not Armstrong")

print("n =", n)
print("s =", s)
```

WAP to find factorial of a number using recursion

```
def fact(n):
  if n==1 or n==0:                    #stopping condition
    return 1
  else:
    return n*fact(n-1)

fact(5)
```

#OPERATOR OVERLOADING

```
class Student():
  def _init_(self,m):
    self.mark=m

  def _add_(self,other):              #magic method
    return self.mark+other.mark

s1=Student(20)
s2=Student(40)
print(s1+s2)                           #s1._add_(s2)
```

```
#A pangram is a sentence that contains every letter of the English alphabet (a–z) at least once.


def pangram(t):
  l=set()
  for char in t.lower():
    if char.isalpha():
      l.add(char)
  return len(l)==26
s="The quick brown fox jumps over the lazy dog"

if pangram(s):
  print("Pangram")
else:
  print("Not a pangram")
```

```
def pangram(t):
  return len(set(i for i in t.lower() if i.isalpha()))==26      #isalpha() is a built-in Python string method used to check whether all characters in a string are alphabetic letters.

s="The quick brown fox jumps over the lazy dog"

if pangram(s):
  print("Pangram")
else:
  print("Not a pangram")
```
